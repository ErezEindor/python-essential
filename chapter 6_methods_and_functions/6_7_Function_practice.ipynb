- LESSER OF TWO EVENS: Write a function that returns the lesser of two given numbers if both numbers are even, but returns the greater if one or both numbers are odd
    - lesser_of_two_evens(2,4) --> 2
    - lesser_of_two_evens(2,5) --> 5

```python
def lesser_of_two_evens (a, b):
    '''this function returns the lesser of 2 given numbers if both numbers are even, but returns the greater if one or both numbers are odd '''
    if a%2 ==0 and b %2 ==0:
        return min(a,b)
    else:
        return max(a,b)
        
```

```python
#check the func 
lesser_of_two_evens(2,4)
```

```python
# check the func 2
lesser_of_two_evens(2,5)
```

## ANIMAL CRACKERS: Write a function takes a two-word string and returns True if both words begin with same letter ##

animal_crackers: ('Levelheaded Lama') --> TRUE

animal_crackers: ('Crazy Kangaroo') --> FALSE

```python
def animal_crackers(text):
    ''' this function takes a two-word string and returns True if both words begin with the same letter
    '''
    words = text.split()
    return words[0][0].lower() == words[1][0].lower()
```

```python
#check the func 1
animal_crackers('Levelheaded Llama')
```

```python
#check the func 2
animal_crackers('Crazy Kangaroo')
```

## MAKES TWENTY: Given two integers, return True if the sum of the integers is 20 or if one of the integers is 20. If not, return False ##

makes_twenty(20,10) --> True

makes_twenty(12,8) --> True

makes_twenty(2,3) --> False

```python
def makes_twenty (n1, n2):
    ''' this function takes two integers and returns True if the sum of the integers is 20 or if one of the integers is 20. If not, it returns False
    '''
    if n1 == 20 or n2 == 20:
        return True
    elif n1 + n2 == 20:
        return True
    else:
        return False 
```

```python
#check the func 1
makes_twenty(20,10)
```

```python
#check the func 2
makes_twenty(12,8)
```

```python
#check the func 3
makes_twenty(2,3)
```
